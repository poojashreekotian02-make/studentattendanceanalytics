tuple in python


1.a tuple is collection of items in python that is ordered,unchangeble(immmutable) and allow duplicate values
2.tuples used to store multiple items in single variable
3.ordered -tuple items have a defined order but that order will not change

```python
#types
#type 1:using parenthesis
nums=(1,2,3,4,5)
colors=("red","green","pink")
mixed=("raj",24,45.6,True)
nested=(23,(45,67),[54,6])
#type 2:without parenthesis
colors="red","green","pink"
print(type(colors))
#type 3:tuple()constructor
values=((1,2,3,4))
print(values,type(values))
list_item=[1,2,3,4,5]
list_tup=tuple(list_item)
#type 4:single value if u not put the comma it will integer 
# if the comma is present then only it will be tuple
num=(45)
print(num,type(num))
num=(45,)
print(num,type(num))
```

```python
colors=("red","green","pink")
#       0      1        2
colors[1]="blue"# we cannot change value once declared bcs tuples are immutable
print(colors)
```

```python
#if u wanna change 1.covert into list
colors=("red","green","pink")
list_colors=list(colors)
list_colors[1]="blue"
colors=tuple(list_colors)
print(colors ,type(colors))
```

tuple packing and unpacking

```python
v1="raj"
v2=23
v3="cse"
#packing 
student=v1,v2,v3
#unpacking
name,age,dept=student
print(name)
print(age)
print(dept)
```

dict functions and methods

```python
student={
    'id':"562",
    'name':"safa",
    'age':88,
    'dept':"cseb"
}
#retrive the single value
get_value=student.get('age')
print(get_value)
#check all the keys
dict_keys=student.keys()
print(dict_keys)
#retrive all the values
dict_values=student.values()
print(dict_values)
#update
update_age=student.update({"age":98})
print(student)
#set default
dep_val=student.setdefault("dept","cs")
print(dep_val)
new_loc=student.setdefault("place","udupi")
print(student)
#copy
c=student.copy()
print(c)
#pop:remove key value based on key
student.pop("dept")
print(student)
#popitem
student.popitem()
print(student)
print(student.items())

```

dict iterate

```python
student={
    'id':"562",
    'name':"safa",
    'age':88,
    'dept':"cseb"
}
#itr over keys
for k in student:
    print(k)
#itr over values
for v in student.values():
    print(v)
#itr over key-values
for k,v in student.items():
    print(k,":",v)
```

Write a script to generate a lookup dictionary where keys represent risk tiers (1 to 15)
and values are then square of each tier number.

```python
dict_square={}
for i in range(1,16):
    dict_square.setdefault(i,i**2)
print(dict_square)
```
