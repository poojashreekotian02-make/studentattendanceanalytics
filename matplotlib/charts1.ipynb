intoduction to matplotlib

```python
import matplotlib.pyplot as plt
import pandas as pd
```

barplot

```python
course_enroll={
    "courses":['java','python','c','c++'],
    "enrolled":[25,15,18,10]
}
course_df=pd.DataFrame(course_enroll)
#create plot
plt.Figure(figsize=(9,5))
bars=plt.bar(course_df["courses"],course_df["enrolled"],color=["#483748","#62624a","#25372B","#6d6a84"],
        edgecolor="pink",width=0.60)
#creating title and level
plt.title("course enrolled student overview",fontsize=15,fontweight='bold',pad=5)
plt.xlabel("course details",fontsize=12)
plt.ylabel("course enrolled student",fontsize=12)
plt.bar_label(bars,padding=2,fontsize=12)
#improve
plt.ylim(0,30)
plt.grid(axis='y',linestyle="--",alpha=0.3)#--,:
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)
plt.tight_layout()
plt.show()
```

lineplot

```python
weekly_sales={
    "days":['mon','tue','wed','thu','fri','sat','sun'],
    'first-week':[35000,30000,38000,32000,39000,42000,45000],
    'second_week':[32000,35000,42000,37000,34000,43000,48000]

}
sales=pd.DataFrame(weekly_sales)
plt.Figure(figsize=(9,5))
#-----first week sales----
plt.plot(sales["days"],sales['first-week'],
         color='red',ls='--',marker='*',label="first week sales")#marker=o,d,*
plt.plot(sales["days"],sales['second_week'],
         color='green',ls='--',marker='*',label="second week sales")
plt.title("shop weekly sales overview",fontsize=15,fontweight='bold')
plt.xlabel("days of weeek",fontsize=12)
plt.ylabel("sales amount per day",fontsize=12)
plt.legend()
plt.tight_layout()
plt.show()
```

scatter plot

```python
student_age={
    
    'age':[20,22,21,20,21,22,23,22,20,24,20,22,21,20,21,22,23,22,20,24],
    'marks':[465,325,625,450,350,490,486,470,250,462,465,325,625,450,350,490,486,470,250,462]
}
std_df=pd.DataFrame(student_age)
plt.Figure(figsize=(9,5))
plt.scatter(std_df['marks'],std_df['age'],
            color='violet',marker="*",s=100,alpha=0.3,label="age/marks")
#create title label
plt.title("based on age check marks",fontsize=15,fontweight='bold')
plt.xlabel("marks",fontsize=12)
plt.ylabel("student age",fontsize=12)
plt.legend()
plt.show
```

pie chart

```python
weekly_sales={
    'Brand':['oneplus','apple','oppo','vivo','samsung'],
    'sales':[23,10,19,32,25]
}
sales_df=pd.DataFrame(weekly_sales)
plt.pie(
    sales_df['sales'],labels=sales_df['Brand'],
    explode=[0,0,0.1,0,0],shadow=True,autopct='%0.2f%%',
    pctdistance=0.7,startangle=30
)
plt.title("weekly sales overview",fontweight='bold')
plt.legend(loc='right')
plt.show()
```
