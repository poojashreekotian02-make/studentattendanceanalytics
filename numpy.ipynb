numpy array and basics

```python
import numpy as np
```

array creation


```python
#1d array creation
arr_1d=np.array([10,20,30,40])
print(arr_1d,type(arr_1d))
#2d array
arr_2d=np.array([[10,20,30],
                 [40,50,60]])
print(arr_2d)
```

```python
arr_2d=np.array([[10,20,30],
                 [40,50,60]
                 ,[50,80]])
print(arr_2d)
```

```python
arr_2d=np.array([[10,20,30],
                 [40,"50k",60]
                 ,[50,80,60]])
print(arr_2d)
```

```python
arr_2d=np.array([[10,20,30],
                 [40,50.00,60]
                 ,[50,80,90]])
print(arr_2d)
```

list v/s numpy array

```python
#ap1
cost_list=[10,20,30,40]
print("selling list",cost_list*2)
cost_arr=np.array([10,20,30,40])
print("selling arr",cost_arr*2)
#ap2
d1_list=[10,20,30,]
d2_list=[20,15,45]
print("total sale",d1_list+d2_list)#not possible
d1_arr=np.array([10,20,30])
d2_arr=np.array([20,15,45])
print("total",d1_arr+d2_arr)
```

```python
import time as t
#list
start=t.time()
list_time=[i*2 for i in range(1000000)]
end=t.time()
print("list operation",end-start)
#numpy array
start=t.time()
arr_time=np.arange(100000)*2
end=t.time()
print("array operation",end-start)
```

creating array from scratch

```python
#zeros
zeros=np.zeros((2,3))
print("zeros matrix:\n",zeros)
#ones
ones=np.ones((3,2))
print("one matrix:\n",ones)
#full
full=np.full((3,3),7)
print("full with num matrix:\n",full)
#random
random=np.random.random((2,2))
print("random matrix(0,1):\n",random)
#sequence start,stop,step
seq=np.arange(10,31,2)
print("sequence of 10 to 30:",seq)
```

vector, matrix and tensor

```python
vector=np.array([10,20,30,40])
print(vector)
matrix=np.array([[10,20,30],
                 [20,50,60]])
print(matrix)
tensor=np.array([
    [[1,2,3],[5,6,4]],
    [[5,6,4],[9,8,8]]
])
print(tensor)
print(tensor.ndim)
```

array properties

```python
arr=np.array([[1,2,3],
              [4,5,6.5]])
print("arr shape",arr.shape)
print("dimension",arr.ndim)
print("size",arr.size)
print("dtype",arr.dtype)#datatype
```

array reshaping


```python
arr=np.arange(12)#it is in the form of array 
print("array:",arr)
reshape=arr.reshape((3,4))#converting array into matrix
print(reshape)
mat=np.array([[10,20,30],
              [40,50,60]])
f1=mat.flatten()#converting back into array
print(f1)
f1=reshape.flatten()#converting back into array
print(f1)
```
