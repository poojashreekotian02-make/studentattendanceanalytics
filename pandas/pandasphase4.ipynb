## groupby merge or join


```python
import pandas as pd
```

groupby

```python
exp=pd.read_excel("subdata_pandas/expense (2).xlsx")
exp.head(5)
```

```python
#total expense for each category
exp.groupby("Category")["Amount"].sum()
exp.groupby("Category")["Amount"].agg(["sum"])
#complete expense summary for each category
exp.groupby("Category")["Amount"].agg(
    ['sum','count','min','max','mean']
)
#how much money u have paid by each payment mode
exp.groupby("Payment Mode")["Amount"].agg(['sum'])
#how many transactions were made using each payment mode
exp.groupby("Payment Mode")["Amount"].agg(['sum','count'])
#for each payment mode,calculate total amount spent across categories
exp.groupby(['Payment Mode','Category'])['Amount'].agg(['sum','count'])
exp.groupby(['Category','Payment Mode'])['Amount'].agg(['sum','count'])
#which categories have highest total expenses
exp.groupby('Category')['Amount'].sum().sort_values(
    ascending=False
)
```

Merge or join

```python
customer=pd.read_csv("subdata_pandas/customer.csv")
customer
payment=pd.read_csv("subdata_pandas/payment.csv")
payment
```

inner join

```python
#inner
pd.merge(left=customer,right=payment,on='customerid',how='inner')
#left
pd.merge(left=customer,right=payment,on='customerid',how='left')
#right
pd.merge(left=customer,right=payment,on='customerid',how='right')
#cross join-task
```
