## numpy functions


```python
import numpy as np
import statistics as st
```

aggregate function

```python
arr=np.array([10,20,30,40,50])
print("sum",np.sum(arr))
print("max",np.max(arr))
print("min",np.min(arr))
print("count",arr.size)
print("comulitive sum",np.cumsum(arr))
print("comulitive product",np.cumprod(arr))
```

## statistical function


```python
sales=np.array([200,180,150,250,170,270,1000,230,500])
print("mean",np.mean(sales))
print("median",np.median(sales))
print("mode",st.mode(sales))
print("standard division",np.std(sales))
```

```python
price=np.array([120,200,180,190,160,170])
sales=np.array([11,12,13,14,16,18])
print(np.corrcoef([price,sales]))
```

```python
sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],  # Paradise Biryani
    [2, 120000, 140000, 160000, 190000],  # Beijing Bites
    [3, 200000, 230000, 260000, 300000],  # Pizza Hub
    [4, 180000, 210000, 240000, 270000],  # Burger Point
    [5, 160000, 185000, 205000, 230000]   # Chai Point
])

# Shape of the data
print("shape", np.shape(sales_data))

# Sample data for first 3 restaurants
print("sample data", sales_data[:3])

# Total sale for each year
print("total sale per year", np.sum(sales_data[:, 1:], axis=0))

# Maximum sale for each restaurant
print("maximum sale per restaurant", np.max(sales_data[:, 1:], axis=1))

# Maximum sale for each year
print("maximum sale every year", np.max(sales_data[:, 1:], axis=0))

# Average sales of each restaurant
print("average sales of every restaurant", np.mean(sales_data[:, 1:], axis=1))

# Cumulative sum of sales
print("cumulative sum of sales", np.cumsum(sales_data[:, 1:]))

# Monthly average sales for all restaurants every year
print("monthly average for all restaurant every year",
      np.mean(sales_data[:, 1:], axis=0) / 12)
```

```python
import numpy as np

# Create two matrices
A = np.array([
    [10, 20],
    [30, 40]
])

B = np.array([
    [2, 4],
    [5, 8]
])

# 1. Addition
print("Addition:")
print(np.add(A, B))


# 2. Subtraction
print("Subtraction:")
print(np.subtract(A, B))


# 3. Multiplication
print("Multiplication:")
print(np.multiply(A, B))


# 4. Division
print("Division:")
print(np.divide(A, B))


# 5. Power
print("Power:")
print(np.power(A, 2))


# 6. Square Root
print("Square Root:")
print(np.sqrt(A))
```

```python
# Add a new row
A = np.array([
    [10, 20],
    [30, 40]
])

new_row = np.array([50, 60])

A = np.vstack((A, new_row))

print("After adding row:")
print(A)
```

```python
A = np.array([
    [10, 20],
    [30, 40],
    [50, 60]
])

# Remove second row
A = np.delete(A, 1, axis=0)

print("After removing row:")
print(A)
```
