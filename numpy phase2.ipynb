numpy array operations

```python
import numpy as np
```

```python
arr=np.array([10,20,30,40,50])
print("basic slicing",arr[:4])
print(" slicing with step",arr[0:5:2])
print("negative index",arr[-1])
```

```python
arr_2d=np.array([[1,2,3,10],
                 [4,5,6,20],
                 [7,8,9,30]])
print("specific element",arr_2d[1,2])
print("entire row:",arr_2d[1:,2:])
print("entire row:",arr_2d[1:,])
print("entire row:",arr_2d[1,:])
print("entire row:",arr_2d[:,1])
print("entire row:",arr_2d[:,1:])
print("last 2 yr sales",arr_2d[2,2:])
print("last 2 yr sales",arr_2d[1,2:])

```

sorting

```python
arr_1d=np.array([1,9,3,4,8,5,7])
print("sorting 1d array",sorted(arr_1d))
print("sorting 1d array",np.sort(arr_1d))#duplicates are allowed
arr_2d=np.array([[2,1,],
                 [3,2],
                 [3,1]])
print("sort 2d array",np.sort(arr_2d))#row wise
#axis=0->col axis=1->row

print("sort 2d array",np.sort(arr_2d,axis=0))#column wise
```

filter

```python
number=np.array([1,9,3,4,8,5,7])
even_number=number[number%2==0]
print(even_number)
```

filter with mask

```python
marks=np.array([67,34,58,89,56,88])
mask_pass=marks>35
print("more than 35(marks)",mask_pass)
```

index

```python
sales=np.array([2500,5600,9600,58000,6400,6300])
idc=[1,3]
res=sales[idc]
print(res)
```

join


```python
arr1=np.array([1,2,3,4])
arr2=np.array([10,20,30,40])
combined=np.concatenate([arr1,arr2])
print("combined",combined)
```

```python
matrix=np.array([[10,20],
                [30,40]])
newmatrix=np.array([[50,60],
                    [70,80]])
print("horizontal\n",np.hstack([matrix,newmatrix]))
print("vertical\n",np.vstack([matrix,newmatrix]))
```

adding and removing

```python
arr=np.array([1,2,3,4,5,6])
print("split",np.split(arr,2))
print("append",np.append(arr,60))
print("insert",np.insert(arr,0,80))
print("remove",np.delete(arr,2))
print("org",arr)
```
