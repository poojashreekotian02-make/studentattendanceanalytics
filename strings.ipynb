string indexing and slicing

```python
#indexing
str="DARKCHOCOLATE"
print("org:",str)
#idx positive
print("5th",str[5])#retrieve the 5 element
print("5th",str[len(str)-1])#when u want to retrive last element
#idx negative
#print last element
print(str[-1])
print(str[-len(str)])#first element
```

```python
#slicing-string_n[start:stop:step]
#default          0     len(s) 1
str="darkchocolate"
print(str[0:4])#dark step=1
print(str[:4])#dark start=0 step=1
print(str[4:len(str)])#chocolate
print(str[0:len(str):2])#step 2 o-2-4-6-8-so on
print(str[4:])#default
print(str[4:13:-1])#i t is not printing anything bcs 4-1 is 3 it will go reverse sooo nothing is printing
#rev_string
rev_s=str[::-1]
print(rev_s)
```

string functions

```python
s="raju told me that he is coder"
#concetnation
print(s + " its real")#adds two strings
print("length",len(s))#to find length of string
print(s.endswith("er"))#is string endswith substring
print(s.replace("he is","i am"))#replace old to new
print(s.find("a"))#first index of substring
print(s.count("e"))#count the frequency of str
print(s.capitalize())#capitalize the fst letter of
print(s.upper())#upper case
print(s.lower())#lowercase
```

```python
#----list functions
color=["red","green","pink"]
#id     0     1        2
print("org",color)
print(len(color))#length
color.append("gold")#insert
print("after adding gold",color)
color[1]="white"#replace
print("after replace",color)
color.insert(2,"pink")#insert at idx position
print("after insert",color)
#remove
color.remove("pink")
print(color)
color.pop()#remove last element
print(color)
```

```python
#looping
color=["red","green","pink"]
#using index access list element
for i in range(len(color)):
    print(i,color[i])
#directly -acces the element
for c in color:
    print(c,end=" ")
```
